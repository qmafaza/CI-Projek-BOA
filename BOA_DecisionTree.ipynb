```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error

def boa_optimize(X_train, y_train, X_test, y_test, n_agents=20, max_iter=30):
    # Parameter BOA
    a = 0.1        # power exponent
    c = 0.01       # sensory modality (akan meningkat tiap iterasi)

    # Hyperparameter search space
    bounds = {
        "max_depth": (2, 20),
        "min_samples_split": (2, 20),
        "min_samples_leaf": (1, 10)
    }

    # fungsi hitung RMSE
    def fitness(params):
        model = DecisionTreeRegressor(
            max_depth=int(params[0]),
            min_samples_split=int(params[1]),
            min_samples_leaf=int(params[2])
        )
        model.fit(X_train, y_train)
        pred = model.predict(X_test)
        return np.sqrt(mean_squared_error(y_test, pred))

    # Inisialisasi populasi
    agents = np.zeros((n_agents, 3))
    for i in range(n_agents):
        agents[i, 0] = np.random.uniform(*bounds["max_depth"])
        agents[i, 1] = np.random.uniform(*bounds["min_samples_split"])
        agents[i, 2] = np.random.uniform(*bounds["min_samples_leaf"])

    # Hitung fitness awal
    fitness_values = np.array([fitness(agent) for agent in agents])
    best_idx = np.argmin(fitness_values)
    best_agent = agents[best_idx].copy()
    best_score = fitness_values[best_idx]

    print(f"Initial best RMSE: {best_score:.4f}")

    # ========== LOOP UTAMA BOA ==========
    for t in range(max_iter):
        fragrance = c * (fitness_values ** (-a))

        for i in range(n_agents):
            r = np.random.rand()

            if r < 0.8:  # global search
                agents[i] = agents[i] + r * (best_agent - agents[i]) * fragrance[i]
            else:        # local search
                j, k = np.random.randint(0, n_agents, 2)
                agents[i] = agents[i] + r * (agents[j] - agents[k]) * fragrance[i]

            # batasan nilai hyperparameter
            agents[i, 0] = np.clip(agents[i, 0], *bounds["max_depth"])
            agents[i, 1] = np.clip(agents[i, 1], *bounds["min_samples_split"])
            agents[i, 2] = np.clip(agents[i, 2], *bounds["min_samples_leaf"])

        # evaluasi
        fitness_values = np.array([fitness(agent) for agent in agents])
        current_best_idx = np.argmin(fitness_values)

        if fitness_values[current_best_idx] < best_score:
            best_score = fitness_values[current_best_idx]
            best_agent = agents[current_best_idx].copy()

        # sensor meningkat setiap iterasi → eksploitasi meningkat
        c = c + 0.01

        print(f"Iterasi {t+1}/{max_iter} | Best RMSE = {best_score:.4f}")

    return best_agent, best_score
```

```python
# Cell tunggal untuk memastikan variabel tersedia, mendefinisikan BOA, dan menjalankan optimasi
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
import os

# 1) Pastikan train_df dan test_df ada; kalau tidak, coba load dari CSV
try:
    train_df  # try reference
    test_df
    print("train_df & test_df sudah ada di memori.")
except NameError:
    print("train_df/test_df tidak ditemukan di memori. Mencoba load 'train_preprocessed.csv' ...")
    if os.path.exists("train_preprocessed.csv"):
        df_all = pd.read_csv("train_preprocessed.csv")
        # jika csv adalah full dataset, split 80/20
        train_df, test_df = train_test_split(df_all, test_size=0.2, random_state=42)
        print("Berhasil load dan split 'train_preprocessed.csv'.")
    else:
        raise FileNotFoundError("Tidak menemukan 'train_preprocessed.csv' di working dir. Pastikan file ada atau jalankan cell preprocessing terlebih dahulu.")

# 2) Pastikan kolom waktu dan fitur tersedia (jika belum diproses, coba extract pickup_datetime)
if 'pickup_datetime' in train_df.columns and not np.issubdtype(train_df['pickup_datetime'].dtype, np.datetime64):
    try:
        train_df['pickup_datetime'] = pd.to_datetime(train_df['pickup_datetime'])
        test_df['pickup_datetime'] = pd.to_datetime(test_df['pickup_datetime'])
    except Exception as e:
        print("Gagal konversi pickup_datetime:", e)

# 3) Buat fitur X/y (distance & passenger_count)
features = ['distance', 'passenger_count']
target = 'fare_amount'

for col in features + [target]:
    if col not in train_df.columns:
        raise KeyError(f"Kolom '{col}' tidak ditemukan di train_df. Pastikan preprocessing menghasilkan kolom ini.")

X_train = train_df[features].copy()
y_train = train_df[target].copy()
X_test = test_df[features].copy()
y_test = test_df[target].copy()

print("Shapes:", X_train.shape, y_train.shape, X_test.shape, y_test.shape)

# 4) Definisi BOA (fungsi)
def boa_optimize(X_train, y_train, X_test, y_test, n_agents=20, max_iter=30, verbose=True):
    a = 0.1
    c = 0.01

    bounds = {
        "max_depth": (2, 20),
        "min_samples_split": (2, 50),  # sedikit diperbesar
        "min_samples_leaf": (1, 20)
    }

    def fitness(params):
        # pastikan integer dan minimal valid
        md = max(1, int(round(params[0])))
        mss = max(2, int(round(params[1])))
        msl = max(1, int(round(params[2])))

        model = DecisionTreeRegressor(
            max_depth=md,
            min_samples_split=mss,
            min_samples_leaf=msl,
            random_state=42
        )
        model.fit(X_train, y_train)
        pred = model.predict(X_test)
        return np.sqrt(mean_squared_error(y_test, pred))

    # inisialisasi agen
    agents = np.zeros((n_agents, 3))
    rng = np.random.default_rng(42)
    for i in range(n_agents):
        agents[i, 0] = rng.uniform(*bounds["max_depth"])
        agents[i, 1] = rng.uniform(*bounds["min_samples_split"])
        agents[i, 2] = rng.uniform(*bounds["min_samples_leaf"])

    fitness_values = np.array([fitness(agent) for agent in agents])
    best_idx = np.argmin(fitness_values)
    best_agent = agents[best_idx].copy()
    best_score = fitness_values[best_idx]

    if verbose:
        print(f"[BOA] Initial best RMSE: {best_score:.4f} | params = {best_agent}")

    # main loop
    for t in range(max_iter):
        # fragrance per agent (hindari div by zero)
        fragrance = c * (fitness_values ** (-a))
        fragrance = np.where(np.isfinite(fragrance), fragrance, c)

        for i in range(n_agents):
            r = rng.random()
            if r < 0.8:
                agents[i] = agents[i] + r * (best_agent - agents[i]) * fragrance[i]
            else:
                j, k = rng.integers(0, n_agents, size=2)
                agents[i] = agents[i] + r * (agents[j] - agents[k]) * fragrance[i]

            # clip ke bounds
            agents[i, 0] = np.clip(agents[i, 0], *bounds["max_depth"])
            agents[i, 1] = np.clip(agents[i, 1], *bounds["min_samples_split"])
            agents[i, 2] = np.clip(agents[i, 2], *bounds["min_samples_leaf"])

        # evaluasi ulang
        fitness_values = np.array([fitness(agent) for agent in agents])
        current_best_idx = np.argmin(fitness_values)
        if fitness_values[current_best_idx] < best_score:
            best_score = fitness_values[current_best_idx]
            best_agent = agents[current_best_idx].copy()

        c = c + 0.01  # sensory modality meningkat
        if verbose:
            print(f"[BOA] Iter {t+1}/{max_iter} | Best RMSE = {best_score:.4f} | best_params = {best_agent}")

    return best_agent, best_score

# 5) Jalankan optimasi BOA
best_params, best_rmse = boa_optimize(X_train, y_train, X_test, y_test, n_agents=20, max_iter=20, verbose=True)

print("\n=== HASIL AKHIR OPTIMASI BOA ===")
print("Best max_depth:", int(round(best_params[0])))
print("Best min_samples_split:", int(round(best_params[1])))
print("Best min_samples_leaf:", int(round(best_params[2])))
print("Best RMSE (validasi test):", best_rmse)

# 6) Latih final model dengan hyperparam terbaik dan evaluasi kembali
final_model = DecisionTreeRegressor(
    max_depth=int(round(best_params[0])),
    min_samples_split=int(round(best_params[1])),
    min_samples_leaf=int(round(best_params[2])),
    random_state=42
)
final_model.fit(X_train, y_train)
pred_final = final_model.predict(X_test)
final_rmse = np.sqrt(mean_squared_error(y_test, pred_final))
print("\nRMSE Decision Tree setelah optimasi BOA:", final_rmse)
```

## BOA Optimization Loop

```python
for t in range(max_iter):
    for i in range(num_butterflies):

        f_i = c * (fitness_values[i] ** a)
        r = np.random.rand()

        if r < p:
            new_pos = population[i] + (r*r * global_best - population[i]) * f_i
        else:
            j, k = np.random.choice(num_butterflies, 2, replace=False)
            new_pos = population[i] + (r*r * population[j] - population[k]) * f_i

        new_pos = np.clip(new_pos, lower_bound, upper_bound)

        new_fit = fitness(new_pos)

        if new_fit < fitness_values[i]:
            population[i], fitness_values[i] = new_pos, new_fit

    best_idx = np.argmin(fitness_values)
    if fitness_values[best_idx] < global_best_fitness:
        global_best = population[best_idx]
        global_best_fitness = fitness_values[best_idx]

    convergence.append(global_best_fitness)

    print(f"Iterasi {t+1}/{max_iter} → Best RMSE: {global_best_fitness:.4f}")

print("\nOptimasi selesai!")
print("Best scale:", global_best[0])
print("Best degree:", int(global_best[1]))
print("Best RMSE:", global_best_fitness)
```
